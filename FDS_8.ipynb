```python
import pandas as pd
import numpy as np

# Creating a DataFrame using a dictionary
data = {
    'Student_ID': [101, 102, 103, 104, 105],
    'Name': ['Ammu', 'Ravi', 'Priya', 'Arjun', 'Sneha'],
    'Department': ['CSE', 'ECE', 'CSE', 'IT', 'ECE'],
    'CGPA': [8.5, 7.8, 9.1, 8.0, 8.7]
}

df = pd.DataFrame(data)

print("Student DataFrame:")
print(df)
```

```python
data = {
    'Marks': [85, 90, 78, 88],
    'Grade': ['A', 'A+', 'B', 'A']
}

index = pd.MultiIndex.from_tuples([
    ('CSE', 101),
    ('CSE', 102),
    ('ECE', 103),
    ('ECE', 104)
], names=['Department', 'Student_ID'])

df = pd.DataFrame(data, index=index)

print(df)

print("\nCSE Students:")
print(df.loc['CSE'])

print("\nStudent 103:")
print(df.loc[('ECE', 103)])
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Ammu', 'Ravi']
})

df2 = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['Priya', 'Arjun']
})

result = pd.concat([df1, df2], axis=0, ignore_index=True)

print(result)
```

```python
df1 = pd.DataFrame({
    'Name': ['Ammu', 'Ravi', 'Priya']
})

df2 = pd.DataFrame({
    'Marks': [90, 85, 95]
})

result = pd.concat([df1, df2], axis=1)

print(result)
```

```python
students = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Name': ['Ammu', 'Ravi', 'Priya', 'Arjun']
})

placements = pd.DataFrame({
    'Student_ID': [101, 103, 105],
    'Company': ['Google', 'Amazon', 'Microsoft']
})

# Inner merge
print("Inner Merge:")
print(pd.merge(students, placements, on='Student_ID', how='inner'))

# Left merge
print("\nLeft Merge:")
print(pd.merge(students, placements, on='Student_ID', how='left'))

# Right merge
print("\nRight Merge:")
print(pd.merge(students, placements, on='Student_ID', how='right'))

# Outer merge
print("\nOuter Merge:")
print(pd.merge(students, placements, on='Student_ID', how='outer'))
```

```python
df = pd.DataFrame({
    'Student': ['Ammu', 'Ravi', 'Priya'],
    'Maths': [90, 80, 95],
    'Science': [85, 88, 92]
})

print("Original DataFrame:")
print(df)

long_df = pd.melt(
    df,
    id_vars=['Student'],
    var_name='Subject',
    value_name='Marks'
)

print("\nReshaped DataFrame:")
print(long_df)
```

```python
df = pd.DataFrame({
    'Maths': [90, 80],
    'Science': [85, 88]
}, index=['Ammu', 'Ravi'])

print("Original DataFrame:")
print(df)

# Convert columns into rows
stacked = df.stack()
print("\nStacked DataFrame:")
print(stacked)

# Convert row levels back into columns
unstacked = stacked.unstack()
print("\nUnstacked DataFrame:")
print(unstacked)
```

```python
df = pd.DataFrame({
    'Student': ['Ammu', 'Ammu', 'Ravi', 'Ravi'],
    'Subject': ['Maths', 'Science', 'Maths', 'Science'],
    'Marks': [90, 85, 80, 88]
})

pivot_df = df.pivot(
    index='Student',
    columns='Subject',
    values='Marks'
)

print(pivot_df)
```

```python
df = pd.DataFrame({
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'CSE'],
    'Gender': ['F', 'M', 'F', 'M', 'F'],
    'Marks': [90, 80, 85, 75, 95]
})

pivot_table = pd.pivot_table(
    df,
    values='Marks',
    index='Department',
    columns='Gender',
    aggfunc='mean'
)

print(pivot_table)
```

```python
df = pd.DataFrame({
    'Name': ['Ammu', 'Ravi', 'Priya', 'Arjun'],
    'Marks': [90, np.nan, 85, np.nan]
})

print("Original DataFrame:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing values with mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print("\nAfter Filling Missing Values:")
print(df)
```

```python
df = pd.DataFrame({
    'ID': [101, 102, 102, 103],
    'Name': ['Ammu', 'Ravi', 'Ravi', 'Priya']
})

print("Original DataFrame:")
print(df)

print("\nDuplicate Rows:")
print(df[df.duplicated()])

df = df.drop_duplicates()

print("\nAfter Removing Duplicates:")
print(df)
```

```python
df = pd.DataFrame({
    'Student_Name': ['Ammu', 'Ravi', 'Priya'],
    'Student_Marks': [85, 95, 75]
})

# Rename columns
df = df.rename(columns={
    'Student_Name': 'Name',
    'Student_Marks': 'Marks'
})

df = df.sort_values(by='Marks', ascending=False)

print(df)
```

```python
df = pd.DataFrame({
    'Name': ['Ammu', 'Ravi', 'Priya'],
    'Marks': [90, 75, 85]
})

df['Result'] = np.where(df['Marks'] >= 80, 'Pass', 'Needs Improvement')

print("After Adding Column:")
print(df)

df = df.drop(columns=['Result'])

print("\nAfter Deleting Column:")
print(df)
```

```python
df = pd.DataFrame({
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'IT'],
    'Marks': [90, 80, 85, 75, 95]
})

result = df.groupby('Department')['Marks'].mean()

print(result)

summary = df.groupby('Department')['Marks'].agg(
    ['mean', 'max', 'min', 'count']
)

print("\nDepartment-wise Summary:")
print(summary)
```

```python
df = pd.DataFrame({
    'Student_ID': [101, 102, 103],
    'Name': ['Ammu', 'Ravi', 'Priya'],
    'Marks': [90, 85, 95]
})

df = df.set_index('Student_ID')
print("After Setting Index:")
print(df)

df = df.reset_index()
print("\nAfter Resetting Index:")
print(df)
```
