```python
import pandas as pd

data1 = {
    'id': ['1','2','3','4','5'],
    'Feature1': ['A','C','E','G','I'],
    'Feature2': ['B','D','F','H','J']
}

df1 = pd.DataFrame(data1)
print(df1)
```

```python
data2 = {
    'id': ['1','2','6','7','8'],
    'Feature1': ['K','M','O','Q','S'],
    'Feature2': ['L','N','P','R','T']
}

df2 = pd.DataFrame(data2)
print(df2)
```

```python
data3 = {
    'id': ['1','2','3','4','5','7','8','9','10','11'],
    'Feature3': [12,13,14,15,16,17,15,12,13,23]
}

df3 = pd.DataFrame(data3)
print(df3)
```

```python
df_row = pd.concat([df1, df2], ignore_index=True)
print(df_row)
```

```python
df_col = pd.concat([df1, df2], axis=1)
print(df_col)
```

```python
df_merge = pd.merge(df1, df3, on='id')
print(df_merge)
```

```python
result = pd.merge(df1, df3, on='id', how='inner')
print(result)
```

```python
result = pd.merge(df1, df3, on='id', how='left')
print(result)
```

```python
result = pd.merge(df1, df3, on='id', how='right')
print(result)
```

```python
result = pd.merge(df1, df3, on='id', how='outer')
print(result)
```

```python
# ============================================
# 11. JOIN
# Join DataFrames using INDEX
# ============================================

emp_details = pd.DataFrame({
    "Name": ["Tanish", "Praneeth", "Ashish"],
    "Dept": ["IT", "HR", "Finance"]
}, index=[101, 102, 103])

emp_salary = pd.DataFrame({
    "Salary": [50000, 60000, 55000]
}, index=[101, 102, 104])

# ============================================

outer_join = emp_details.join(
    emp_salary,
    how="outer"
)

print("\n===== OUTER JOIN =====")
print(outer_join)
```

```python
result = pd.merge(
    df1,
    df3,
    on='id',
    how='outer',
    indicator=True
)

print(result)
```

```python
result = pd.merge(
    df1,
    df3,
    left_on='id',
    right_on='id',
    how='inner'
)

print(result)
```
